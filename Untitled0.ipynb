```python
from ultralytics import YOLO

print("Ultralytics ready")
```

```python
model = YOLO("yolov8n.pt")
```

```python
results = model.predict(
    source="/content/busy_airport.jpg",
    conf=0.00,
    save=True
)

result = results[0]
```

```python
results = model.track(
    source="farm.mp4",
    tracker="bytetrack.yaml",
    conf=0.35,
    show=False,
    save=True,
    stream=True
)
```

```python
import cv2
video_path = "/content/farm.mp4"
cap = cv2.VideoCapture(video_path)

for frame_num in range(457, 461):
    # OpenCV frame indexing starts at 0
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_num - 1)
    success, frame = cap.read()

    if not success:
        print(f"Could not read frame {frame_num}")
        continue

    results = model.predict(frame, conf=0.35, verbose=False)
    annotated = results[0].plot()
```
